```python
import os

from abc import ABC, abstractmethod
# Constantes
OPERADORES_VALIDOS = ["+", "-", "*", "/", "!"]

def limpiar_pantalla():
    os.system("cls" if os.name == "nt" else "clear")
    
def solicitar_numero(mensaje):
    while True:
        try:
            return float(input(mensaje))  
        except ValueError:
            limpiar_pantalla()
            print("❌ Debe ingresar un número válido.")

def solicitar_operador():
    while True:
        operador = input(f"Ingrese operador {', '.join(OPERADORES_VALIDOS)}: ")
        if operador in OPERADORES_VALIDOS:
            return operador
        print("❌ Operador no válido.")
    
def solicitar_mayor_que(mensaje, mayor_que=0):
    while True:
        numero = solicitar_numero(mensaje)
        if numero >= mayor_que:
             return numero
        else:
            limpiar_pantalla()
            print(f"❌ Debe ingresar un número válido y mayor o igual a {mayor_que}.") 
```

```python
class CalculadoraAbstracta(ABC):
    @abstractmethod
    def tipo_calculadora(self):
        pass
    
class Calculadora(CalculadoraAbstracta):
    def __init__(self):
        self.__numero1 = 0
        self.__numero2 = 0
    
    def set_numero1 (self, numero1):
        self.__numero1 = numero1
        
    def set_numero2 (self, numero2):
        self.__numero2 = numero2
        
    def get_numero1(self):
        return self.__numero1
    
    def get_numero2(self):
        return self.__numero2
    
    def sumar(self):
        return self.__numero1 + self.__numero2
    
    def restar(self):
        return self.__numero1 - self.__numero2
    
    def multiplicar(self):
        return self.__numero1 * self.__numero2

    def dividir(self):
        try:
            return self.__numero1 / self.__numero2
        except ZeroDivisionError:
            print("Error: El denominador debe ser diferente a cero (0)")
    
    def tipo_calculadora(self):
        return "Calculadora Básica"

class CalculadoraFactorial(Calculadora):
    
    def __init__(self):
        super().__init__()
        
    def set_numero(self, numero):
        while True:
            if numero < 0:
                numero = solicitar_numero("Ingrese un número mayor o igual a 0: ")
            else:
                break
        super().set_numero1(numero)
    
    def factorial(self):
        numero = super().get_numero1()
        print(numero)
        return self._factorial_aux(int(numero))
        
    def _factorial_aux(self, numero):
        if numero == 0 or numero == 1:  # Caso base
            return 1
        return numero * self._factorial_aux(numero - 1) 

    def tipo_calculadora(self):
        return "Calculadora Factorial"
```

```python
def main():
    # Programa principal
    
    while True:
        numero1 = solicitar_numero("Ingrese el primer número: ")
        operador = solicitar_operador()

        if operador != "!":
            calc = Calculadora()
            calc.set_numero1(numero1)
            numero2 = solicitar_numero("Ingrese el segundo número: ")
            calc.set_numero2(numero2)
        
        else:
            calc = CalculadoraFactorial()
            calc.set_numero(numero1)

        match operador:
            case '+':
                resultado = calc.sumar()
            case '-':
                resultado = calc.restar()
            case '*':
                resultado = calc.multiplicar()
            case '/':
                resultado = calc.dividir()
            case '!':
                resultado = calc.factorial()
                
        print(f"✅ Resultado del operador '{operador}' es: {resultado}", flush=True)
        salir = input("¿Desea realizar otra operación (s/n)?").lower() or 's'
        print(calc.tipo_calculadora())

        if not salir == 's':
            print("Finaliza el programa.!!")
            break
```

```python
main()
```
