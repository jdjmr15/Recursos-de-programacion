# Calculadora en Python con match-case
Utilizamos la comunicacion con el usuario utilizando la consola usando la funcion `input()` aplicando la conversion de tipos de datos y condicionales utilice `match` y `case` en python.

Los alumnos debe refactorizar este codigo usando `if`, `elif` y `else`.


```python
num1 = float(input("Ingrese el primer número: "))
operador = input("Ingrese el operador (+, -, *, /): ")
num2 = float(input("Ingrese el segundo número: "))
```

```python
operacion = ""
resultado = 0
if operador == '+':
    operacion = "suma"
    resultado = num1 + num2
elif operador == '-':
    operacion = "resta"
    resultado = num1 - num2
elif operador == '*':
    operacion = "multiplicación"
    resultado = num1 * num2
elif operador == '/':
    operacion = "división"
    resultado = num1 / num2
else:
    operacion = "error"
```

```python
print("El numero ingresado es:", num1)
print("El numero ingresado es:", num2)
print("El operador ingresado es:", operador)
print(f"El resultado de la {operacion} es:", resultado)
```
